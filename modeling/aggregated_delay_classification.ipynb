```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import sqlalchemy as sqla
```

## DATABASE CONNECTION

```python
DB_USER = "postgres"
DB_PASSWORD = "postgres"
DB_HOST = "localhost"
DB_PORT = "5432"
DB_NAME = "postgres"

connection_string = (
    f"postgresql+psycopg2://{DB_USER}:{DB_PASSWORD}@"
    f"{DB_HOST}:{DB_PORT}/{DB_NAME}"
)

engine = sqla.create_engine(connection_string)

try:
    with engine.connect() as connection:
        print("Connected to PostgreSQL successfully!")
except Exception as error:
    print(f"Error: {error}")

connection = engine.connect()
```

## Import Data

```python
query = """
SELECT
w.date,
EXTRACT(HOUR FROM w.hour) AS hour_of_day,
AVG(w.temperature) AS avg_temperature,
AVG(w.dewpoint) AS avg_dewpoint,
AVG(w.relative_humidity) AS avg_relative_humidity,
AVG(w.precipitation) AS avg_precipitation,
AVG(w.snowfall) AS avg_snowfall,
AVG(w.wind_direction) AS avg_wind_direction,
AVG(w.wind_speed) AS avg_wind_speed,
AVG(w.pressure) AS avg_pressure
FROM WEATHER w
GROUP BY w.date, hour_of_day
ORDER BY w.date, hour_of_day;
"""

weather = pd.read_sql(sqla.text(query), connection)

query = """
    SELECT
        t.departure_date AS date,
        EXTRACT(HOUR FROM t.real_time_departure) AS hour_of_day,
        AVG(t.delay_arrival) AS avg_delay_arrival,
        AVG(t.delay_departure) AS avg_delay_departure
    FROM TRAIN_DATA t
    GROUP BY t.departure_date, hour_of_day
    ORDER BY t.departure_date, hour_of_day;
"""

train_data = pd.read_sql(sqla.text(query), connection)

query = """
    SELECT * FROM TYPE_DAY;
"""

type_day = pd.read_sql(sqla.text(query), connection)


delay_weather_data = train_data.merge(
    weather, on=["date", "hour_of_day"], how="inner"
)

delay_weather_data = type_day.merge(
    delay_weather_data, on=["date"], how="inner"
)

delay_weather_data["date"] = pd.to_datetime(delay_weather_data["date"])

delay_weather_data["date"] = delay_weather_data["date"].dt.dayofyear

delay_weather_data["day_sin"] = np.sin(
    2 * np.pi * delay_weather_data["date"] / 365
)
delay_weather_data["day_cos"] = np.cos(
    2 * np.pi * delay_weather_data["date"] / 365
)

# Drop the original 'date' column
delay_weather_data.drop(columns=["date"], inplace=True)

print(delay_weather_data.shape)
```

## Testing with Classifier

```python
from sklearn.ensemble import (
    AdaBoostClassifier,
    BaggingClassifier,
    ExtraTreesClassifier,
    GradientBoostingClassifier,
    RandomForestClassifier,
)
from sklearn.linear_model import (
    LogisticRegression,
    RidgeClassifier,
    SGDClassifier,
)
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
)
from sklearn.model_selection import KFold, train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
```

```python
def train_and_evaluate_model(
    training_features,
    test_features,
    training_targets,
    test_targets,
    model,
    print_results=True,
):
    model.fit(training_features, training_targets)
    predicted_targets = model.predict(test_features)
    print("Model", model)
    print("Accuracy", accuracy_score(test_targets, predicted_targets))
    if print_results:
        print(
            "Confusion Matrix\n",
            confusion_matrix(test_targets, predicted_targets),
        )
        print(
            "Classification Report\n",
            classification_report(test_targets, predicted_targets),
        )
```

```python
input_values = delay_weather_data[
    [
        "day_sin",
        "day_cos",
        "holiday",
        "weekend",
        "day_after_rest",
        "hour_of_day",
    ]
]
raw_date_features = raw_date_data[
    ["date", "holiday", "weekend", "day_after_rest", "hour_of_day"]
]
target_values = delay_weather_data["avg_delay_departure"]

median_y = target_values.median()

delay_classes = target_values.apply(
    lambda input_values: 1 if input_values > median_y else 0
)

cyclic_date_accuracies = []
raw_date_accuracies = []

for i in range(20):
    training_features, test_features, training_targets, test_targets = (
        train_test_split(
            input_values, delay_classes, test_size=0.2, random_state=i
        )
    )
    (
        raw_date_training_features,
        raw_date_test_features,
        raw_date_training_targets,
        raw_date_test_targets,
    ) = train_test_split(
        raw_date_features, delay_classes, test_size=0.2, random_state=i
    )
    model = RandomForestClassifier()
    model.fit(training_features, training_targets)
    predicted_targets = model.predict(test_features)
    cyclic_date_accuracies.append(
        accuracy_score(test_targets, predicted_targets)
    )
    model = RandomForestClassifier()
    model.fit(raw_date_training_features, raw_date_training_targets)
    predicted_targets = model.predict(raw_date_test_features)
    raw_date_accuracies.append(
        accuracy_score(raw_date_test_targets, predicted_targets)
    )

print("Accuracy with date sin and cos", np.mean(cyclic_date_accuracies))
print("Accuracy without date sin and cos", np.mean(raw_date_accuracies))
```

## Hyperparameter Tuning

```python
# On RandomForest

input_values = delay_weather_data[
    [
        "avg_temperature",
        "avg_dewpoint",
        "avg_relative_humidity",
        "avg_precipitation",
        "avg_snowfall",
        "avg_wind_direction",
        "avg_wind_speed",
        "avg_pressure",
        "day_sin",
        "day_cos",
        "holiday",
        "weekend",
        "day_after_rest",
        "hour_of_day",
    ]
]
target_values = delay_weather_data["avg_delay_departure"]

target_values = target_values.apply(
    lambda input_values: 1 if input_values < 180 else 0
)


training_features, test_features, training_targets, test_targets = (
    train_test_split(
        input_values, target_values, test_size=0.3, random_state=42
    )
)

n_estimators = [10, 25, 50, 75, 100, 150]
max_depths = [2, 5, 10, 20, 50, 100, None]
methods = ["gini", "entropy", "log_loss"]

best_score = 0
best_params = None
for n_estimator in n_estimators:
    for max_depth in max_depths:
        for method in methods:
            random_forest = RandomForestClassifier(
                n_estimators=n_estimator, max_depth=max_depth, criterion=method
            )
            random_forest.fit(training_features, training_targets)
            predicted_targets = random_forest.predict(test_features)
            score = accuracy_score(test_targets, predicted_targets)
            if score > best_score:
                best_score = score
                best_params = (n_estimator, max_depth, method)


print("Best score:", best_score)
print("Best params:", best_params)
```
