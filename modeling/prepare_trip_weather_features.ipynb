```python
query = """
SELECT * FROM stations;
"""

stations = pd.read_sql(sqla.text(query), connection)

province_centroids = [
    (0, "Bruxelles", 4.367414882050033, 50.83642580516115),
    (1, "Antwerpen", 4.721043779498864, 51.231900712244745),
    (2, "Limburg", 5.4357209224443475, 50.98831159244354),
    (3, "Oost-Vlaanderen", 3.8188605810442007, 51.03629347952689),
    (4, "Vlaams Brabant", 4.59072072950776, 50.87303943005528),
    (5, "West-Vlaanderen", 3.0620446668608117, 51.01019740435232),
    (6, "Brabant Wallon", 4.589736973914637, 50.66605965845173),
    (7, "Hainaut", 3.9681364929631004, 50.46387637972707),
    (8, "Liège", 5.7372263688290746, 50.518876085615986),
    (9, "Luxembourg", 5.516735682549194, 49.95849015160492),
    (10, "Namur", 4.862346452029854, 50.252686235186815),
]

province_data = pd.DataFrame(
    province_centroids,
    columns=["province_id", "province_name", "province_long", "province_lat"],
)


def haversine(start_longitude, start_latitude, end_longitude, end_latitude):
    earth_radius_km = 6371
    start_longitude, start_latitude, end_longitude, end_latitude = map(
        np.radians,
        [start_longitude, start_latitude, end_longitude, end_latitude],
    )
    longitude_delta = end_longitude - start_longitude
    latitude_delta = end_latitude - start_latitude

    haversine_term = (
        np.sin(latitude_delta / 2) ** 2
        + np.cos(start_latitude)
        * np.cos(end_latitude)
        * np.sin(longitude_delta / 2) ** 2
    )
    central_angle = 2 * np.arcsin(np.sqrt(haversine_term))
    return earth_radius_km * central_angle


def find_nearest_province(station_row):
    distances = haversine(
        station_row["longitude"],
        station_row["latitude"],
        province_data["province_long"].values,
        province_data["province_lat"].values,
    )
    nearest_province_index = distances.argmin()
    return province_data.iloc[nearest_province_index]


# Apply the function
nearest_provinces = stations.apply(find_nearest_province, axis=1)

# Concatenate station info with nearest province
stations_with_province = pd.concat(
    [
        stations.reset_index(drop=True),
        nearest_provinces.reset_index(drop=True),
    ],
    axis=1,
)

route_station_ids = [
    159,
    151,
    149,
    150,
    156,
    157,
    154,
    245,
    773,
    132,
    310,
    365,
    422,
    300,
    628,
    614,
    591,
    540,
    118,
    178,
    235,
    281,
    480,
    108,
    642,
    625,
    623,
    557,
    558,
    384,
    199,
    559,
    645,
    123,
    195,
    73,
    261,
    560,
    187,
    438,
    172,
    342,
    341,
    80,
    503,
    629,
    267,
    313,
    608,
    339,
    452,
    564,
    493,
    317,
    704,
    757,
    70,
]

filtered_stations = stations_with_province[
    stations_with_province["id"].isin(route_station_ids)
]
filtered_stations = filtered_stations.drop(
    columns=[
        "province_long",
        "province_lat",
        "region",
        "latitude",
        "longitude",
        "geom",
        "name",
        "province_name",
    ]
)
filtered_stations["id"] = pd.Categorical(
    filtered_stations["id"], categories=route_station_ids, ordered=True
)
filtered_stations = filtered_stations.sort_values("id").reset_index(drop=True)

print(filtered_stations.head())
```

```python
DB_USER = "postgres"
DB_PASSWORD = "postgres"
DB_HOST = "localhost"
DB_PORT = "5432"
DB_NAME = "postgres"

connection_string = (
    f"postgresql+psycopg2://{DB_USER}:{DB_PASSWORD}@"
    f"{DB_HOST}:{DB_PORT}/{DB_NAME}"
)

# Créer un moteur SQLAlchemy
engine = sqla.create_engine(connection_string)

try:
    with engine.connect() as connection:
        print("Connected to PostgreSQL successfully!")
except Exception as error:
    print(f"Error: {error}")

connection = engine.connect()

query = """
SELECT * FROM stations;
"""

stations = pd.read_sql(sqla.text(query), connection)

province_centroids = [
    (0, "Bruxelles", 4.367414882050033, 50.83642580516115),
    (1, "Antwerpen", 4.721043779498864, 51.231900712244745),
    (2, "Limburg", 5.4357209224443475, 50.98831159244354),
    (3, "Oost-Vlaanderen", 3.8188605810442007, 51.03629347952689),
    (4, "Vlaams Brabant", 4.59072072950776, 50.87303943005528),
    (5, "West-Vlaanderen", 3.0620446668608117, 51.01019740435232),
    (6, "Brabant Wallon", 4.589736973914637, 50.66605965845173),
    (7, "Hainaut", 3.9681364929631004, 50.46387637972707),
    (8, "Liège", 5.7372263688290746, 50.518876085615986),
    (9, "Luxembourg", 5.516735682549194, 49.95849015160492),
    (10, "Namur", 4.862346452029854, 50.252686235186815),
]

province_data = pd.DataFrame(
    province_centroids,
    columns=["province_id", "province_name", "province_long", "province_lat"],
)


# Apply the function
nearest_provinces = stations.apply(find_nearest_province, axis=1)

# Concatenate station info with nearest province
stations_with_province = pd.concat(
    [
        stations.reset_index(drop=True),
        nearest_provinces.reset_index(drop=True),
    ],
    axis=1,
)

route_station_ids = [
    159,
    151,
    149,
    150,
    156,
    157,
    154,
    245,
    773,
    132,
    310,
    365,
    422,
    300,
    628,
    614,
    591,
    540,
    118,
    178,
    235,
    281,
    480,
    108,
    642,
    625,
    623,
    557,
    558,
    384,
    199,
    559,
    645,
    123,
    195,
    73,
    261,
    560,
    187,
    438,
    172,
    342,
    341,
    80,
    503,
    629,
    267,
    313,
    608,
    339,
    452,
    564,
    493,
    317,
    704,
    757,
    70,
]

filtered_stations = stations_with_province[
    stations_with_province["id"].isin(route_station_ids)
]
filtered_stations = filtered_stations.drop(
    columns=[
        "province_long",
        "province_lat",
        "region",
        "latitude",
        "longitude",
        "geom",
        "name",
        "province_name",
    ]
)
filtered_stations["id"] = pd.Categorical(
    filtered_stations["id"], categories=route_station_ids, ordered=True
)
filtered_stations = filtered_stations.sort_values("id").reset_index(drop=True)

query = """
SELECT
    td.id,
    td.departure_date,
    td.train_number,
    td.relation,
    td.train_service,
    td.ptcar_number,
    td.line_number_departure,
    td.real_time_arrival,
    td.real_time_departure,
    td.planned_time_arrival,
    td.planned_time_departure,
    td.delay_arrival,
    td.delay_departure,
    ptcar.id AS ptcar_id,
    ptcar.name AS name_travel,
    td.line_number_arrival,
    dep.name AS departure_station_name,
    arr.name AS arrival_station_name,
    ptcar.longitude AS ptcar_longitude,
    ptcar.latitude AS ptcar_latitude
FROM train_data td
JOIN STATIONS ptcar ON td.ptcar_name = ptcar.id
JOIN STATIONS dep ON td.station_departure = dep.id
JOIN STATIONS arr ON td.station_arrival = arr.id
WHERE td.station_arrival = '487'
  AND td.station_departure = '159'
ORDER BY td.id;
"""

route_data = pd.read_sql(sqla.text(query), connection)
route_data = route_data.drop(columns=["train_service", "relation"])
print("Road:", len(route_station_ids))

current_trip_rows = []
complete_trip_rows = []
route_stop_index = 0

for i, row in route_data.iterrows():
    ptcar_id = row["ptcar_id"]
    if ptcar_id == route_station_ids[route_stop_index]:
        current_trip_rows.append(row)
        route_stop_index += 1
        if len(current_trip_rows) == len(route_station_ids):
            complete_trip_rows.extend(current_trip_rows)
            current_trip_rows = []
            route_stop_index = 0
    else:
        current_trip_rows = []
        route_stop_index = 0

complete_trips = pd.DataFrame(complete_trip_rows)

query = """
    SELECT * FROM TYPE_DAY;
"""


type_day = pd.read_sql(sqla.text(query), connection)
complete_trips = complete_trips.drop(
    columns=[
        "departure_station_name",
        "arrival_station_name",
        "id",
        "name_travel",
        "ptcar_number",
        "ptcar_id",
    ]
)
calendar_enriched_trips = pd.merge(
    complete_trips,
    type_day,
    left_on="departure_date",
    right_on="date",
    how="left",
)
calendar_enriched_trips.loc[
    calendar_enriched_trips.index[::57], "delay_arrival"
] = 0

calendar_enriched_trips.loc[
    calendar_enriched_trips.index[::57], "real_time_arrival"
] = calendar_enriched_trips.loc[
    calendar_enriched_trips.index[::57], "real_time_departure"
]
calendar_enriched_trips.loc[
    calendar_enriched_trips.index[::57], "planned_time_arrival"
] = calendar_enriched_trips.loc[
    calendar_enriched_trips.index[::57], "planned_time_departure"
]
calendar_enriched_trips.loc[
    calendar_enriched_trips.index[::57], "line_number_arrival"
] = calendar_enriched_trips.loc[
    calendar_enriched_trips.index[::57], "line_number_departure"
]


calendar_enriched_trips.loc[
    calendar_enriched_trips.index[56::57], "line_number_departure"
] = 162
calendar_enriched_trips.loc[
    calendar_enriched_trips.index[56::57], "line_number_arrival"
] = 162

last_stop_indices = calendar_enriched_trips.index[56::57]
calendar_enriched_trips.loc[last_stop_indices, "real_time_departure"] = (
    calendar_enriched_trips.loc[last_stop_indices, "real_time_arrival"]
)

calendar_enriched_trips.loc[last_stop_indices, "planned_time_departure"] = (
    calendar_enriched_trips.loc[last_stop_indices, "planned_time_arrival"]
)

calendar_enriched_trips.loc[last_stop_indices, "delay_departure"] = (
    calendar_enriched_trips.loc[last_stop_indices, "delay_arrival"]
)

station_ids = filtered_stations["province_id"].values
total_rows = len(calendar_enriched_trips)
repeated_ids = np.tile(station_ids, total_rows // len(station_ids) + 1)[
    :total_rows
]

repeated_ids_df = pd.DataFrame({"repeated_station_id": repeated_ids})
calendar_enriched_trips = pd.concat(
    [calendar_enriched_trips, repeated_ids_df], axis=1
)
query = """
SELECT
    w.date,
    EXTRACT(HOUR FROM w.hour) AS hour_of_day,
    w.temperature AS temperature,
    w.dewpoint AS dewpoint,
    w.relative_humidity AS relative_humidity,
    w.precipitation AS precipitation,
    w.snowfall AS snowfall,
    w.wind_direction AS wind_direction,
    w.wind_speed AS wind_speed,
    w.pressure AS pressure,
    w.province AS province_id
FROM WEATHER w
GROUP BY
    w.date,
    EXTRACT(HOUR FROM w.hour),
    w.temperature,
    w.dewpoint,
    w.relative_humidity,
    w.precipitation,
    w.snowfall,
    w.wind_direction,
    w.wind_speed,
    w.pressure,
    w.province
ORDER BY w.date, hour_of_day;

"""

weather = pd.read_sql(sqla.text(query), connection)

calendar_enriched_trips["time"] = pd.to_datetime(
    calendar_enriched_trips["planned_time_arrival"], format="%H:%M:%S"
)
calendar_enriched_trips["hour"] = calendar_enriched_trips["time"].dt.hour

trip_features = pd.merge(
    calendar_enriched_trips,
    weather,
    left_on=["departure_date", "hour", "repeated_station_id"],
    right_on=["date", "hour_of_day", "province_id"],
    how="left",
)
trip_features = trip_features.drop(
    columns=[
        "date_x",
        "date_y",
        "hour_of_day",
        "time",
        "hour",
        "repeated_station_id",
    ]
)
print("Middle of processing")


def parse_time_to_seconds(time_value):
    if pd.isna(time_value):
        return 0
    try:
        if isinstance(time_value, (int, float)):
            return int(time_value)
        if str(time_value).isdigit():
            return int(time_value)
        parsed_time = pd.to_datetime(
            time_value, format="%H:%M:%S", errors="coerce"
        )
        if pd.isna(parsed_time):
            return 0
        return (
            parsed_time.hour * 3600
            + parsed_time.minute * 60
            + parsed_time.second
        )
    except Exception:
        return 0


trip_features["departure_date"] = pd.to_datetime(
    trip_features["departure_date"], format="%Y-%m-%d"
)

trip_features["departure_year"] = trip_features["departure_date"].dt.year
trip_features["departure_month"] = trip_features["departure_date"].dt.month
trip_features["departure_day"] = trip_features["departure_date"].dt.day
trip_features["departure_weekday"] = trip_features["departure_date"].dt.weekday

trip_features["real_time_arrival"] = trip_features["real_time_arrival"].apply(
    parse_time_to_seconds
)
trip_features["planned_time_arrival"] = trip_features[
    "planned_time_arrival"
].apply(parse_time_to_seconds)
trip_features["planned_time_departure"] = trip_features[
    "planned_time_departure"
].apply(parse_time_to_seconds)
trip_features["real_time_departure"] = trip_features[
    "real_time_departure"
].apply(parse_time_to_seconds)

trip_features = trip_features.drop(
    columns=[
        "departure_date",
        "line_number_departure",
        "line_number_arrival",
        "train_number",
    ]
)
trip_features["delay_arrival"] = trip_features["delay_arrival"].apply(
    parse_time_to_seconds
)

valid_trips = []

for i in range(0, len(trip_features), 57):
    trip = trip_features.iloc[i: i + 57]
    if trip.shape[0] < 57:
        print("Not enough data for this trajet, skipping")
        continue

    if trip.isnull().values.any():
        continue

    if trip["delay_arrival"].abs().max() > 36000:
        print("Delay arrival exceeds 10000 seconds, skipping")
        continue

    valid_trips.append(trip)

valid_trip_features = pd.concat(valid_trips, ignore_index=True)
valid_trip_features["stop_index"] = valid_trip_features.index % 57
```
