```python
import holidays
import pandas as pd

# Generate date range from 2014-01-01 to 2024-12-31
date_range = pd.date_range(start="2014-01-01", end="2024-12-31")
calendar_data = pd.DataFrame(date_range, columns=["date"])

calendar_data["holiday"] = 0

school_holidays = [
    ("01-01-2014", "05-01-2014"),
    ("03-03-2014", "09-03-2014"),
    ("07-04-2014", "20-04-2014"),
    ("01-07-2014", "31-08-2014"),
    ("27-10-2014", "02-11-2014"),
    ("22-12-2014", "04-01-2015"),
    ("16-02-2015", "22-02-2015"),
    ("06-04-2015", "19-04-2015"),
    ("01-07-2015", "31-08-2015"),
    ("02-11-2015", "08-11-2015"),
    ("21-12-2015", "03-01-2016"),
    ("08-02-2016", "14-02-2016"),
    ("28-03-2016", "10-04-2016"),
    ("01-07-2016", "31-08-2016"),
    ("31-10-2016", "06-11-2016"),
    ("26-12-2016", "08-01-2017"),
    ("27-02-2017", "05-03-2017"),
    ("03-04-2017", "16-04-2017"),
    ("01-07-2017", "31-08-2017"),
    ("30-10-2017", "05-11-2017"),
    ("25-12-2017", "07-01-2018"),
    ("12-02-2018", "18-02-2018"),
    ("02-04-2018", "15-04-2018"),
    ("01-07-2018", "31-08-2018"),
    ("29-10-2018", "04-11-2018"),
    ("24-12-2018", "06-01-2019"),
    ("04-03-2019", "10-03-2019"),
    ("08-04-2019", "21-04-2019"),
    ("01-07-2019", "31-08-2019"),
    ("28-10-2019", "03-11-2019"),
    ("23-12-2019", "05-01-2020"),
    ("24-02-2020", "01-03-2020"),
    ("06-04-2020", "19-04-2020"),
    ("01-07-2020", "31-08-2020"),
    ("02-11-2020", "08-11-2020"),
    ("23-12-2020", "03-01-2021"),
    ("15-02-2021", "21-02-2021"),
    ("05-04-2021", "18-04-2021"),
    ("01-07-2021", "31-08-2021"),
    ("01-11-2021", "07-11-2021"),
    ("27-12-2021", "09-01-2022"),
    ("28-02-2022", "06-03-2022"),
    ("04-04-2022", "17-04-2022"),
    ("01-07-2022", "28-08-2022"),
    ("24-10-2022", "06-11-2022"),
    ("26-12-2022", "08-01-2023"),
    ("20-02-2023", "05-03-2023"),
    ("01-05-2023", "14-05-2023"),
    ("08-07-2023", "27-08-2023"),
    ("23-10-2023", "05-11-2023"),
    ("25-12-2023", "07-01-2024"),
    ("26-02-2024", "10-03-2024"),
    ("29-04-2024", "12-05-2024"),
    ("06-07-2024", "25-08-2024"),
    ("21-10-2024", "03-11-2024"),
    ("23-12-2024", "31-12-2024"),
]

school_holiday_dates = set()
for start, end in school_holidays:
    school_holiday_dates.update(
        pd.date_range(
            start=pd.to_datetime(start, dayfirst=True),
            end=pd.to_datetime(end, dayfirst=True),
        )
    )
calendar_data.loc[
    calendar_data["date"].isin(school_holiday_dates), "holiday"
] = 1  # For school holidays

holiday_dict = {}
for year in range(2014, 2025):
    belgian_public_holidays = holidays.BE(years=year, language="fr")
    holiday_dict.update(belgian_public_holidays)

public_holidays = set(holiday_dict.keys())

calendar_data.loc[calendar_data["date"].dt.date.isin(
    public_holidays), "holiday"] = 2

calendar_data["weekend"] = (
    calendar_data["date"].dt.dayofweek >= 5
).astype(int)

rest_days = (calendar_data["holiday"] == 2) | (calendar_data["weekend"] == 1)
calendar_data["day_after_rest"] = (
    ~rest_days & rest_days.shift(1, fill_value=False)
).astype(int)

# Save to CSV
file_path = "../data/reference/dates_2014_2024.csv"
calendar_data.to_csv(file_path, index=False)
```
