```python
import pandas as pd
```

```python
train_relations = pd.read_csv("../data/reference/RELATION.csv")

train_relations["type"] = 0
train_relations.loc[
    train_relations["RELATION"].str.startswith("S", na=False), "type"
] = 1
train_relations.loc[
    train_relations["RELATION"].str.startswith("P", na=False), "type"
] = 2
train_relations.loc[
    train_relations["RELATION"].str.startswith("L", na=False), "type"
] = 3
train_relations.loc[
    train_relations["RELATION"].str.startswith("IC", na=False), "type"
] = 4
train_relations.loc[
    train_relations["RELATION"].str.startswith("EXTRA", na=False), "type"
] = 5
train_relations.loc[
    train_relations["RELATION"].str.startswith("ICE", na=False), "type"
] = 6
train_relations.loc[
    train_relations["RELATION"].str.startswith("T", na=False), "type"
] = 6
train_relations.loc[
    train_relations["RELATION"].str.startswith("INT", na=False), "type"
] = 6
train_relations.loc[
    train_relations["RELATION"].str.startswith("ICT", na=False), "type"
] = 7

train_relations.to_csv("../data/reference/RELATION.csv", index=False)
```
