<span style="color:green">This file is used to map the stations to their respective regions, along with the longitude and latitude of each station and store it in the file stations_with_regions.csv </span>


```python
import time

import pandas as pd

stations = pd.read_csv("../data/reference/UNIQUE_STATION.csv")

print(stations.head())
```

```python
from geopy.geocoders import Nominatim
from tqdm import tqdm

geolocator = Nominatim(user_agent="geo_region_finder")


def get_region(location_name):
    location = geolocator.geocode(
        location_name, exactly_one=True, addressdetails=True
    )

    if location and "address" in location.raw:
        address = location.raw["address"]
        # Get the most relevant administrative region
        if address.get("country_code") == "be":
            region = address.get(
                "region",
                address.get("region", address.get("county", "Unknown region")),
            )
            return region
        else:
            return "Unknown region"
    return "Unknown region"


tqdm.pandas()  # Enable progress bar
stations["REGION"] = stations["STATION"].progress_apply(
    lambda input_values: (time.sleep(1), get_region(input_values))[1]
)
```

```python
station_coordinates = pd.read_csv("../data/geo/coordinates.csv", sep=";")

station_coordinates = station_coordinates.melt(
    id_vars=["Geo Point"],
    value_vars=[
        "Abréviation LST FR courte",
        "Abréviation LST NL courte",
        "Abréviation LST FR complète",
        "Abréviation LST NL complète",
        "Nom FR court",
        "Nom NL court",
        "Nom FR moyen",
        "Nom NL moyen",
        "Nom FR complet",
        "Nom NL complet",
    ],
    var_name="source_column",
    value_name="key_column",
)
station_coordinates.to_csv(
    "../data/interim/station_name_coordinates.csv", index=False
)

stations = pd.merge(
    stations,
    station_coordinates,
    left_on="STATION",
    right_on="key_column",
    how="left",
)
stations = stations.drop(columns=["key_column", "source_column"])
stations = stations.drop_duplicates(keep="first")


stations[["Latitude", "Longitude"]] = stations["Geo Point"].str.split(
    ", ", expand=True
)

stations[["Latitude", "Longitude"]] = stations[
    ["Latitude", "Longitude"]
].astype(float)

stations = stations.drop(columns=["Geo Point"])

stations.to_csv(
    "../data/interim/station_coordinates_uncorrected.csv", index=False
)
```
